```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rows = int(input("Enter number of layers (rows): "))
cols = int(input("Enter number of sites per layer (columns): "))
t_max = int(input("Enter total number of time steps (T): "))
```

```python
species_labels = ['P', 'K', 'S']
fractions = [0.3, 0.3, 0.4]  
etch_times = {'P': 1, 'K': 1, 'S': 3}
```

```python

np.random.seed(42)
lattice = np.random.choice(species_labels, size=(rows, cols), p=fractions)
```

```python
def plot_lattice(lattice, title=""):
    fig, ax = plt.subplots(figsize=(min(12, cols), 3))
    ax.set_xlim(-0.5, lattice.shape[1] - 0.5)
    ax.set_ylim(-lattice.shape[0] + 0.5, 0.5)
    ax.set_xticks(range(lattice.shape[1]))
    ax.set_yticks(-np.arange(lattice.shape[0]))
    ax.grid(True, which='both', color='lightgrey')
    ax.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
    for n in range(lattice.shape[0]):
        for m in range(lattice.shape[1]):
            c = lattice[n, m]
            label = c if c != '' else ''
            ax.text(m, -n, label, ha='center', va='center', fontsize=16,
                    bbox=dict(boxstyle="circle", ec='black', fc='white'))
    ax.set_title(title, fontsize=16)
    plt.show()
```

```python

def etch_lattice(prev_lattice, t_now):
    new_lattice = prev_lattice.copy()
    prev_state = prev_lattice.copy()  # Use this to check exposure correctly
    for layer in range(rows):
        for col in range(cols):
            species = prev_state[layer, col]
            if species == '':
                continue
            # Exposure check: All layers above must be empty
            exposed = all(prev_state[above, col] == '' for above in range(layer))
            if exposed and (t_now - layer == etch_times[species]):
                new_lattice[layer, col] = ''
    return new_lattice
```

```python
lattices = [lattice]
plot_lattice(lattice, title="t = 0")

for t in range(1, t_max + 1):
    new_lattice = etch_lattice(lattices[-1], t_now=t)
    lattices.append(new_lattice)
    plot_lattice(new_lattice, title=f"t = {t}T")
```
